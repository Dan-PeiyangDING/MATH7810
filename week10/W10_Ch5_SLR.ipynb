## 1. The question

One row is one purchase. `time_on_page` is in minutes. `order_value` is in dollars.

Among these purchases, how does the predicted order value change when time on the page increases by one minute?

The population line is

$$y = \beta_0 + \beta_1 x + \varepsilon$$

From the data we fit $\hat{y} = b_0 + b_1 x$. The slope $b_1$ is an estimated regression coefficient.


## 2. Load the purchases

**Done when:** the printout shows 800 rows and the columns `time_on_page` and `order_value`.


```python
# One row is one purchase. Both columns are observed.
# CSV is in the data folder next to this notebook.
import pandas as pd

purchases = pd.read_csv("data/checkout_purchases.csv")
print("rows, columns:", purchases.shape)
print(purchases.head())
```

## 3. Scatter and the fitted line

The line is fit with `scipy.stats.linregress`. The printed slope is dollars of predicted order value per extra minute on the page.


```python
import matplotlib.pyplot as plt
from scipy import stats

slope, intercept, r_value, p_value, se_slope = stats.linregress(
    purchases["time_on_page"], purchases["order_value"]
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(purchases["time_on_page"], purchases["order_value"], alpha=0.6)
x_line = [purchases["time_on_page"].min(), purchases["time_on_page"].max()]
ax.plot(x_line, [intercept + slope * x for x in x_line], color="C1")
ax.set_title("Order value and time on page")
ax.set_xlabel("Time on page (minutes)")
ax.set_ylabel("Order value ($)")
plt.tight_layout()
plt.show()

print(f"intercept = {intercept:.3f}")
print(f"slope = {slope:.3f}")
print(f"r = {r_value:.3f}")
print(
    f"Each extra minute on the page is associated with about {slope:.2f} dollars "
    "higher predicted order value. This is an association, not a cause."
)
```
