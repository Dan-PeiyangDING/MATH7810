# 1. Lists

A list stores an **ordered** sequence of values (items can have different types).


```python
# Purpose: create lists and check their type

cheeses = ['Cheddar', 'Edam', 'Gouda']
number = [17, 123]
empty = []
mixed = ['abc', 2.0, True, [10, 20]]  # a list can hold mixed types, even another list

print(cheeses, number, empty)
print(type(cheeses), type(mixed))
```

### Indexing and slicing

Python counts from **0**. Slicing `t[start:stop]` goes up to, but **not including**, `stop`.


```python
# Purpose: indexing — get one element

print(cheeses[0])   # first
print(cheeses[-1])  # last
print('Edam' in cheeses)  # membership test
```

```python
# Purpose: slicing — get a sublist (a new list)

t = ['a', 'b', 'c', 'd', 'e', 'f']
print(t[1:3])  # 'b', 'c'
print(t[:4])   # first four
print(t[3:])   # from index 3 to the end
print(t)       # original list unchanged
```

### Change a list; length; a short loop


```python
# Purpose: modify by index, append, and check length

numbers = [17, 123]
numbers[1] = 5          # replace the second element
print(numbers)

numbers.append(99)      # add at the end
print(numbers)

print('length:', len(numbers))
```

```python
# Purpose: update every element with a for loop

values = [17, 5]
for i in range(len(values)):
    values[i] = values[i] * 2
print(values)
```

```python
# Purpose: list comprehension — same idea, shorter syntax

values = [17, 5]
values = [x * 2 for x in values]
print(values)
```

# 2. Dictionaries

A dictionary maps **keys** to **values**. A Pandas **Series** can be thought of the same way: it maps **index** labels to data values. A **DataFrame** is often built from a dict of lists (column name → column of values).


```python
# Purpose: why a dict helps — two lists side by side make lookup inconvenient
# Lookup takes two steps; the lists can stop lining up; it is slow if the list is long

eng = ['one', 'two', 'three']
sp = ['uno', 'dos', 'tres']

# To translate 'two', find its position in eng, then use the same position in sp
i = eng.index('two')
print(sp[i])
```

```python
# Purpose: dictionary lookup by key
eng2sp = {'one': 'uno', 'two': 'dos', 'three': 'tres'}
print(eng2sp['two'])
print('one' in eng2sp)   # checks keys, not values
print('uno' in eng2sp)   # False — 'uno' is a value
print(len(eng2sp))
```

```python
# Purpose: loop over keys (and read each value)
counts = {'chuck': 1, 'annie': 42, 'jan': 100}
for name in counts:
    print(name, counts[name])
```

### Bridge to Pandas (Week 4)

A **dict of equal-length lists** is a natural table: each key is a column name.


```python
# Purpose: dict of lists — each key is a column name (how a DataFrame is often created)

sales = {
    'store': ['A', 'A', 'B', 'B'],
    'day': ['Mon', 'Tue', 'Mon', 'Tue'],
    'revenue': [120, 150, 90, 110],
}
print(sales)
print('revenue column:', sales['revenue'])
# Next week: pd.DataFrame(sales)
```

# 3. Set

A set stores **unique** unordered items — useful for “distinct categories.”


```python
# Purpose: duplicates are dropped in a set

payment = ['Card', 'Cash', 'Card', 'Mobile', 'Cash']
print(set(payment))
print('Card' in set(payment))
```

# 4. Tuple

A tuple is like a list but **immutable** (you cannot assign to an element).


```python
# Purpose: create a tuple; index it; show immutability

t = ('a', 'b', 'c')
print(type(t), t[0], t[1:3])

# t[0] = 'A'  # would raise TypeError — uncomment to see
```
