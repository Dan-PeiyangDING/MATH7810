## Learning goals
- Create 1D/2D NumPy arrays
- Indexing, slicing, and vectorized operations
- Use aggregations (`mean`, `std`, …) as **NumPy syntax** (stats meaning was Week 2)


```python
# Purpose: load NumPy

import numpy as np
```

### Create arrays


```python
# Purpose: create 1D and 2D arrays

a = np.array([1, 2, 3, 4, 5])  # 1D array from a list
print(a, a.dtype, a.shape)     # values, data type, and length of each axis
```

The comma in (5,) is required Python: (5) is just the number 5; (5,) is a tuple of length 1. NumPy uses that to show “one axis, not two.”

```python
zeros = np.zeros(5)           # 1D array of five 0s
ones = np.ones((2, 3))        # 2×3 array of 1s
seq = np.arange(0, 10, 2)     # 0, 2, 4, 6, 8 (start, stop, step)
print(zeros)                  # display zeros
print(ones)                   # display the 2×3 ones
print(seq)                    # display the even numbers
```

### Indexing and slicing


```python
# Purpose: indexing and slicing

print(a[0], a[-1])
print(a[1:4])
M = np.array([[10,20,30],[40,50,60]])
print(M[0,1])
print(M[:,1])  # second column
```

### Aggregations
These are the same summaries as Week 2. Here we observe **NumPy syntax**: one method on the whole array (`x.mean()`), not a Python loop. A **list** has no `.mean()` and cannot be compared to 4 in one step.


```python
# Purpose: aggregations as NumPy syntax (mean, SD, min, max, count)

wait_list = [2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0]
print(sum(wait_list) / len(wait_list))  # mean by hand — a list has no .mean()
# wait_list.mean()   # AttributeError — uncomment to see
# wait_list > 4      # TypeError — uncomment to see

x = np.array(wait_list)  # same values, now an array
print(x.mean(), x.std(ddof=1), x.min(), x.max())
print(x.sum(), (x > 4).sum())  # count how many values are greater than 4
```

### Vectorized operations


```python
# Purpose: vectorized operations (no loop)

print(a * 2)
print(a + np.array([10,10,10,10,10]))
```
